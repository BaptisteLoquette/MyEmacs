# Attention Visualization — BertViz

This notebook demonstrates **BertViz** to inspect self-attention in a transformer encoder.

We show two views:
1. **head_view**: attention links for a single input sequence
2. **model_view**: aggregated attention across all layers and heads

```python
from transformers import AutoTokenizer, AutoModel
from bertviz import head_view, model_view

# Load a small pretrained encoder
model_name = 'distilbert-base-uncased'
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModel.from_pretrained(model_name, output_attentions=True)

sentence = "The transformer architecture revolutionized natural language processing."
inputs = tokenizer(sentence, return_tensors='pt')
outputs = model(**inputs)
attention = outputs.attentions  # tuple of (batch, heads, seq, seq) for each layer
tokens = tokenizer.convert_ids_to_tokens(inputs['input_ids'][0])
print('Tokens:', tokens[:12])
```

```python
# Head view: interactive attention links for selected layer/head
head_view(attention, tokens)
```

The **head view** above is interactive: hover over a token to see incoming attention weights. Colors encode attention strength.

```python
# Model view: all layers and heads in a grid
model_view(attention, tokens)
```

The **model view** shows the full attention landscape. You can toggle layers and heads to compare specialization (e.g., positional vs semantic heads).

```python
# Interactive sentence selector with ipywidgets
from ipywidgets import interact, Textarea

def visualize_attention(text="The cat sat on the mat."):
    inputs = tokenizer(text, return_tensors='pt')
    outputs = model(**inputs)
    toks = tokenizer.convert_ids_to_tokens(inputs['input_ids'][0])
    return head_view(outputs.attentions, toks)

interact(visualize_attention, text=Textarea(value='The cat sat on the mat.', description='Sentence'))
```
